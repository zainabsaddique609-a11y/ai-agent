```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Hum yahan ek clean aur asaan dataset (Student Performance) load kar rahe hain
url = "https://raw.githubusercontent.com/ageron/handson-ml/master/datasets/housing/housing.csv"
df = pd.read_csv(url)

print("Dataset kamyabi se load ho gaya! Total rows aur columns:")
print(df.shape)
df.head()
```

```python
# Check karte hain ke koi value khali (missing) toh nahi hai
print("Missing values:")
print(df.isnull().sum())

# Data ke basic numbers (average, min, max) nikalte hain
df.describe()
```

```python
# Missing values ko theek kar dete hain taake graphs mein koi masla na ho
df['total_bedrooms'].fillna(df['total_bedrooms'].median(), inplace=True)

# Ab hum alag-alag graphs banate hain
plt.figure(figsize=(12, 10))

# 1. Histogram (House Age ka)
plt.subplot(2, 2, 1)
sns.histplot(df['housing_median_age'], bins=30, kde=True, color='purple')
plt.title('1. Housing Median Age Distribution')

# 2. Scatter Plot (Rooms vs Population)
plt.subplot(2, 2, 2)
sns.scatterplot(x='total_rooms', y='population', data=df, color='teal')
plt.title('2. Total Rooms vs Population')

# 3. Box Plot (Median House Value)
plt.subplot(2, 2, 3)
sns.boxplot(y=df['median_house_value'], color='orange')
plt.title('3. Median House Value Boxplot')

# 4. Correlation Heatmap
plt.subplot(2, 2, 4)
sns.heatmap(df.select_dtypes(include=[np.number]).corr(), annot=False, cmap='coolwarm')
plt.title('4. Correlation Heatmap')

plt.tight_layout()
plt.show()
```

```python
# Baqi ke graphs banate hain
plt.figure(figsize=(12, 10))

# 5. Line Chart / Trend
plt.subplot(2, 2, 1)
plt.plot(df['housing_median_age'].head(50), color='green', marker='o')
plt.title('5. Housing Age Trend (First 50 Rows)')

# 6. Bar Chart (Average House Value by Age groups)
plt.subplot(2, 2, 2)
df['age_group'] = pd.cut(df['housing_median_age'], bins=5, labels=['New', 'Moderate', 'Old', 'Very Old', 'Ancient'])
df.groupby('age_group')['median_house_value'].mean().plot(kind='bar', color='coral')
plt.title('6. Avg House Value by Age Group')

# 7. Distribution Plot of Median Income
plt.subplot(2, 2, 3)
sns.kdeplot(df['median_income'], fill=True, color='purple')
plt.title('7. Median Income Density Distribution')

# 8. Scatter Plot with Income vs House Value
plt.subplot(2, 2, 4)
sns.scatterplot(x='median_income', y='median_house_value', data=df.sample(500), color='brown', alpha=0.5)
plt.title('8. Median Income vs House Value')

plt.tight_layout()
plt.show()
```

```python
# Aakhri 2 graphs aur insights ke liye
plt.figure(figsize=(12, 5))

# 9. Count Plot of Age Groups
plt.subplot(1, 2, 1)
sns.countplot(x=df['age_group'], palette='viridis')
plt.title('9. Count of Houses in Each Age Group')
plt.xticks(rotation=45)

# 10. Violin Plot
plt.subplot(1, 2, 2)
sns.violinplot(x=df['housing_median_age'] < 30, y=df['median_house_value'], palette='muted')
plt.title('10. House Value based on Age Split')

plt.tight_layout()
plt.show()

# Kuch basic insights jo aap apni report mein likh sakti hain:
print("--- MAJOR INSIGHTS ---")
print(1, "Median house value kaafi had tak median income ke sath barhti hai.")
print(2, "Most houses ki median age 15 se 35 saal ke darmiyan hai.")
print(3, "Total rooms aur population ka aapas mein strong positive link hai.")
```
